```python
from ultralytics import YOLO

# Confirm installation
import ultralytics
print("Ultralytics YOLO version:", ultralytics.__version__)
```

```python
# Options: yolov11n.pt (nano), yolov11s.pt (small), yolov11m.pt (medium), yolov11l.pt (large), yolov11x.pt (extra large)
model = YOLO("yolo11s.pt")  # Start from pre-trained small model
```

```python
import torch
print(torch.cuda.is_available())      # should be True
print(torch.cuda.get_device_name(0))  # should print your GPU name# Shows your GPU name
```

```python
# Train with your dataset
results = model.train(
    data="data.yaml",   # Path to your data.yaml
    epochs=50,         # Number of epochs
    imgsz=640,          # Image size (default 640)
    batch=16,           # Adjust batch size based on your GPU memory
    workers=4,          # Number of dataloader workers
    device='0',        # CUDA device (0 for first GPU)
    name="yolo11_custom" # Folder name where training results will be saved
)
```

```python
metrics = model.val()  # Run validation
print(metrics)
```

```python
# Example inference on test folder
results = model.predict(source="./test/images", save=True, conf=0.25)

# Example inference on a single image
results = model.predict(source="test\images\point03_95_png_jpg.rf.e4d03d0dcc5bc8f2d0e9cb845125207f.jpg", show=True)
```
