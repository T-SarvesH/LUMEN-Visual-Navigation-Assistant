```python
#Importing the libraries

from roboflow import Roboflow
from dotenv import load_dotenv
import os
from ultralytics import YOLO
```

```python
load_dotenv()
```

```python
#Importing the API key from the .env in root

Rflow_key = os.getenv("ROBOFLOW_API_KEY")
print("Roboflow API Key loaded response:", Rflow_key is not None)
```

```python
#Accessing the project and the dataset

rf = Roboflow(api_key=Rflow_key)
project = rf.workspace("lumen-visual-assistant").project("vehicle-aorar")

print(f"Project load Response: {project is not None}")
```

```python
#Exporting the dataset

dataset = project.version(2).download("yolov8")
print(f"Dataset creation Response: {dataset is not None}")
```

```python
vhcl_dtn_dir = os.environ.get("VHCL_DTN_PATH")
print(f"Loaded the Vehicle Detection Dir path: {vhcl_dtn_dir is not None}")
```

```python
#Create a new dir for result saving

save_dir = vhcl_dtn_dir + "/" + "Version1-Results"
os.makedirs(save_dir, exist_ok=True)
```

```python
model = YOLO('yolo11s.pt')

results = model.train(
    data=vhcl_dtn_dir + '/' + 'Vehicle-2' + '/' + 'data.yaml',     
    epochs=50,               
    imgsz=640,               
    project=save_dir,               
    device="0",          
    amp=True,
)
```
